## Tìm số chiều PCA rồi ghi quy tắc trước huấn luyện

Tìm số chiều nhỏ nhất đạt 90% cho từng development snapshot; lấy **giá trị lớn nhất của 60 tháng**
làm số chiều cố định chung. Như vậy mọi tháng giữ ít nhất 90%. Không mặc định dùng 4 chiều cũ;
không thay số chiều từng tháng sau freeze. Scaler/PCA vẫn fit riêng trong từng snapshot.
Cell tính dưới đây chỉ đọc. Cell tiếp theo ghi bảng evidence, freeze và config trước khi K-Means chạy.


```python
selection_rows = []
for p in PREPARED:
    vectors, _ = preprocess(p['rows'], dict(features=FEATURES, winsor_quantile=0,
        scaling='robust_per_snapshot', reduction={'method':'none'}))
    _, full_pca = fit_pca(vectors, len(FEATURES))
    ratios = np.array(full_pca['explained_variance_ratio'])
    cumulative = np.cumsum(ratios)
    assert np.isfinite(ratios).all() and cumulative[-1] >= THRESHOLD
    selection_rows.append(dict(snapshot_date=p['snapshot_date'], n_eligible=p['eligible_count'],
        minimum_components_90=int(np.searchsorted(cumulative,THRESHOLD))+1,
        **{f'PC{i+1}_explained_variance':float(v) for i,v in enumerate(ratios)},
        **{f'PC{i+1}_cumulative':float(v) for i,v in enumerate(cumulative)}))
selection = pd.DataFrame(selection_rows)
N_COMPONENTS = int(selection.minimum_components_90.max())
selection['n_components_frozen'] = N_COMPONENTS
selection['cumulative_explained_variance_frozen'] = selection[f'PC{N_COMPONENTS}_cumulative']
assert selection.cumulative_explained_variance_frozen.ge(THRESHOLD).all()
if N_COMPONENTS > 1:
    assert selection[f'PC{N_COMPONENTS-1}_cumulative'].lt(THRESHOLD).any()
display(selection[['snapshot_date','n_eligible','minimum_components_90',
                   'n_components_frozen','cumulative_explained_variance_frozen']])
print('Số chiều cố định từ dữ liệu mới:', N_COMPONENTS)
```

```python
# CELL ĐẦU TIÊN GHI FILE; chưa fit K-Means.
require_new_targets()
FREEZE_DIR.mkdir(parents=True,exist_ok=False)
selection.to_csv(FREEZE_DIR/'pca_selection_diagnostics.csv',index=False,lineterminator='\n')
write_rows(FREEZE_DIR/'pca_selection_diagnostics.jsonl', selection.to_dict('records'))
FREEZE = dict(artifact_id='m2-task6-pca-freeze-v2',created_at=now(),scope='development_only',
    development_snapshots=DEV_DATES, features=FEATURES, scaling='robust_per_snapshot', global_k=2,
    rule='smallest_uniform_count_covering_every_development_snapshot',
    explained_variance_threshold=THRESHOLD,n_components=N_COMPONENTS, source_hashes=HASHES,
    evidence_hashes={p.name:sha(p) for p in sorted(FREEZE_DIR.iterdir())})
write_json(FREEZE_DIR/'freeze.json', FREEZE)
CONFIG = copy.deepcopy(MASTER)
CONFIG.update(stage='M2_TASK_3_GLOBAL_K_FROZEN',run_id=RUN_ID,output_dir=rel(TASK_DIR))
CONFIG['clustering'].update(algorithm='pca_kmeans',k=2,**PARAMS)
CONFIG['clustering']['pca'].update(component_rule_status='frozen_on_development',
    n_components=N_COMPONENTS, component_rule=FREEZE['rule'],
    explained_variance_threshold=THRESHOLD,freeze_evidence=rel(FREEZE_DIR/'freeze.json'))
CONFIG['clustering']['reduction'] = dict(method='pca',n_components=N_COMPONENTS)
CONFIG['assumptions']['pca_components'] = (
    f'Số chiều chung {N_COMPONENTS} là max số chiều tối thiểu đạt 90% trên 60 development snapshots; '
    f'evidence: {rel(FREEZE_DIR/"freeze.json")}; không dùng holdout.')
validate_protocol(CONFIG)
write_json(CONFIG_PATH,CONFIG)
CONFIG_HASH, FREEZE_HASH = sha(CONFIG_PATH),sha(FREEZE_DIR/'freeze.json')
print('Đã ghi config:',CONFIG_PATH)
print('Đã ghi freeze:',FREEZE_DIR/'freeze.json')
```

## Huấn luyện thật rồi kiểm chứng model

Cell kế tiếp gọi runner để fit lại từng tháng. K và tham số K-Means lấy từ quyết định đã phê duyệt.
Runner tính diagnostics K=2…8 theo `k_range`; **assignment chỉ dùng K=2**, không chọn lại K sau PCA.
Silhouette trong runner Python có thể chạy lâu. Chờ cell kết thúc, rồi chạy tiếp các cell kiểm chứng.
Trạng thái task6 chỉ trở thành `complete` sau cell xuất cuối; lỗi hoặc ngắt giữa chừng giữ trạng thái chờ xuất.


```python
assert not TASK_DIR.exists(), 'Không chạy đè artifact.'
if MODEL_DIR.exists():
    assert MODEL_DIR.is_dir() and not any(MODEL_DIR.iterdir()), 'Model mirror đã có nội dung.'
assert sha(CONFIG_PATH) == CONFIG_HASH and sha(FREEZE_DIR/'freeze.json') == FREEZE_HASH
assert sha(DATA_DIR/'manifest.json') == HASHES['c8_manifest_sha256']
assert sha(DATA_DIR/'canonical/feature_snapshots.jsonl') == HASHES['c8_feature_sha256']
TARGET, MANIFEST = experiment(DATA_DIR,CONFIG_PATH,ROOT)
assert TARGET.resolve() == TASK_DIR.resolve()
if MANIFEST['status'] != 'complete':
    raise RuntimeError(f'Runner thất bại: {MANIFEST.get("error")}; xem {TASK_DIR/"manifest.json"}.')
MANIFEST.update(status='exporting',task6_exports_complete=False)
write_json(TASK_DIR/'manifest.json',MANIFEST)
assert MANIFEST['n_snapshots'] == 60 and MANIFEST['n_assignments'] == EXPECTED_ASSIGNMENTS

```

```python
assert MANIFEST['status'] == 'exporting' and MANIFEST['config'] == CONFIG
assert MANIFEST['source_manifest_hash'] == HASHES['c8_manifest_sha256']
assignments = pd.read_csv(TASK_DIR/'assignments.csv')
assert len(assignments) == EXPECTED_ASSIGNMENTS and set(assignments.snapshot_date) == set(DEV_DATES)
assert not assignments.duplicated(['snapshot_date','security_id']).any()
diagnostics = pd.read_csv(TASK_DIR/'diagnostics.csv')
selected = diagnostics[diagnostics.k.eq(2)].copy()
assert len(selected) == 60 and selected.snapshot_date.is_unique and selected.converged.eq(True).all()
assert np.isfinite(selected[['silhouette','davies_bouldin','inertia','cluster_balance']]).all().all()
stored = {(r['snapshot_date'],r['raw_cluster_id']):r for r in read_rows(TASK_DIR/'profiles.jsonl')}
assert len(stored) == 120
scores, loadings, pca_rows, scalers, profiles, centroids, inputs = [],[],[],[],[],[],[]
for p in PREPARED:
    day, rows = p['snapshot_date'],p['rows']
    model = read_json(TASK_DIR/f'models/{day}.json')
    assert model['config'] == m2_clustering_config(CONFIG)
    scaler, red = model['scaler'],model['scaler']['_reduction']
    assert red['n_components'] == N_COMPONENTS and red['fit_scope'] == 'supplied_snapshot_only'
    X = np.array([[r[f] for f in FEATURES] for r in rows],dtype=float)
    center,scale = np.array([scaler[f]['center'] for f in FEATURES]),np.array([scaler[f]['scale'] for f in FEATURES])
    q25,q75 = np.quantile(X,[.25,.75],axis=0)
    assert np.allclose(center,np.median(X,axis=0)) and np.allclose(scale,np.where(q75==q25,1.,q75-q25))
    fit_time = datetime.fromisoformat(model['fit_as_of'])
    assert fit_time.date().isoformat() == day
    assert all(datetime.fromisoformat(r['available_at']) <= fit_time for r in rows)
    Z = ((X-center)/scale-np.array(red['input_center'])) @ np.array(red['components']).T
    centers = np.array(model['centroids'])
    assert centers.shape == (2,N_COMPONENTS)
    labels = ((Z[:,None,:]-centers[None,:,:])**2).sum(axis=2).argmin(axis=1)
    saved = assignments[assignments.snapshot_date.eq(day)].set_index('security_id')
    assert set(saved.index) == {r['security_id'] for r in rows}
    assert labels.tolist() == saved.loc[[r['security_id'] for r in rows],'raw_cluster_id'].tolist()
    diag = selected[selected.snapshot_date.eq(day)].iloc[0]
    assert np.isclose(((Z-centers[labels])**2).sum(),diag.inertia)
    ratios = np.array(red['explained_variance_ratio'])
    expected = selection[selection.snapshot_date.eq(day)].iloc[0]
    assert np.allclose(ratios,[expected[f'PC{i+1}_explained_variance'] for i in range(N_COMPONENTS)])
    assert ratios.sum() >= THRESHOLD
    pca_rows.append(dict(snapshot_date=day,n_eligible=len(rows),n_components=N_COMPONENTS,
        minimum_components_90=int(expected.minimum_components_90),cumulative_explained_variance=float(ratios.sum()),
        **{f'PC{i+1}_explained_variance':float(v) for i,v in enumerate(ratios)}))
    for r,z in zip(rows,Z):
        scores.append(dict(snapshot_date=day,security_id=r['security_id'],**{f'PC{i+1}':float(v) for i,v in enumerate(z)}))
        inputs.append(dict(snapshot_date=day,security_id=r['security_id'],available_at=r['available_at'],
                           data_version=r['data_version'],**{f:r[f] for f in FEATURES}))
    for i,component in enumerate(red['components'],1):
        loadings.extend(dict(snapshot_date=day,component=f'PC{i}',feature=f,loading=float(v)) for f,v in zip(FEATURES,component))
    scalers.extend(dict(snapshot_date=day,feature=f,center=float(c),scale=float(s),
        lower=scaler[f]['lower'],upper=scaler[f]['upper']) for f,c,s in zip(FEATURES,center,scale))
    for cluster in range(2):
        mask = labels == cluster
        mean = X[mask].mean(axis=0)
        profile = stored[(day,cluster)]
        assert profile['size'] == int(mask.sum()) and np.allclose(mean,[profile['centroid'][f] for f in FEATURES])
        assert np.allclose(centers[cluster],Z[mask].mean(axis=0))
        aligned = saved.loc[[r['security_id'] for r,flag in zip(rows,mask) if flag],'aligned_cluster_id']
        assert aligned.nunique() == 1 and int(aligned.iloc[0]) == profile['aligned_cluster_id']
        profiles.append(dict(snapshot_date=day,raw_cluster_id=cluster,aligned_cluster_id=profile['aligned_cluster_id'],
            size=int(mask.sum()),size_ratio=float(mask.mean()),**dict(zip(FEATURES,mean.tolist()))))
        centroids.append(dict(snapshot_date=day,raw_cluster_id=cluster,aligned_cluster_id=profile['aligned_cluster_id'],
            **{f'PC{i+1}':float(v) for i,v in enumerate(centers[cluster])}))
    assert sha(MODEL_DIR/f'{day}.json') == sha(TASK_DIR/f'models/{day}.json')
profile_frame,pca_summary = pd.DataFrame(profiles),pd.DataFrame(pca_rows)
assert profile_frame.columns.tolist() == ['snapshot_date','raw_cluster_id','aligned_cluster_id','size','size_ratio',*FEATURES]
assert len(scores) == len(inputs) == EXPECTED_ASSIGNMENTS and len(profiles) == 120
assert len(loadings) == 60*N_COMPONENTS*8 and len(scalers) == 60*8
assert pca_summary.minimum_components_90.max() == N_COMPONENTS
print('PASS: scaler/PCA, assignments, inertia, centroids, profiles và model mirror.')
```
